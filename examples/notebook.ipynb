# stdplot in a Jupyter notebook

Inside a Jupyter kernel, `Display.AUTO` shows each figure inline, below its cell, instead of opening a Qt window. This needs the notebook extra:

```
pip install "stdplot[notebook]"
```

````python
import numpy as np

import stdplot

frequency = np.linspace(0.0, 50.0, 1000)
````

## Inline SVG

The default preview target is an SVG of the figure at the size it was drawn (8 x 6 in, unless the call passes `size=(width, height)`), so the inline figure stays sharp at any zoom.

````python
with stdplot.figure("resonance") as (fig, ax):
    ratio = frequency / 12.0
    for damping in (0.1, 0.3, 0.7):
        gain = 1.0 / np.sqrt((1.0 - ratio**2) ** 2 + (2.0 * damping * ratio) ** 2)
        ax.semilogy(frequency, gain, label=rf"$\zeta = {damping}$")
    ax.set_xlabel(r"Frequency $f$ [Hz]")
    ax.set_ylabel(r"Gain $|H(f)|$")
    ax.legend(title="Damping ratio")
````

## Inline PNG

The preview target sets the inline format and resolution, and optionally its size. A PNG preview keeps the notebook light when a figure has many points. A `Plotter` binds the config to every figure drawn with it.

````python
png_preview = stdplot.Plotter(
    stdplot.Config(
        preview=stdplot.Target(stdplot.Format.PNG, stdplot.Size(6.0, 4.0), dpi=150),
    ),
)

rng = np.random.default_rng(seed=0)
samples = rng.normal(size=(20_000, 2)) @ np.array([[1.0, 0.0], [0.6, 0.8]])

with png_preview.figure("correlated_samples") as (fig, ax):
    ax.plot(samples[:, 0], samples[:, 1], ".", markersize=1, alpha=0.3)
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_aspect("equal")
````

Saving works as in a script: set `output_dir` in the config, then pass `save=True` once to the `Plotter`, or to a single call, as in `export.py`.
