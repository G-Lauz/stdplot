# stdplot in a Jupyter notebook

Inside a Jupyter kernel, `Display.AUTO` shows each figure inline, below its cell, instead of opening a Qt window. This needs the notebook extra:

```
pip install "stdplot[notebook]"
```

````python
import numpy

from stdplot import Config, Format, Size, Stdplot, Target

stdplot = Stdplot()

time = numpy.linspace(0.0, 10.0, 500)
displacement = numpy.exp(-0.3 * time) * numpy.cos(2.0 * numpy.pi * 0.8 * time)
````

## Inline SVG

The default preview target is an SVG of the figure at the size it was drawn (8 x 6 in, unless the call passes `size=(width, height)`), so the inline figure stays sharp at any zoom.

````python
with stdplot.figure("damped_oscillation") as (fig, ax):
    ax.plot(time, displacement)
    ax.set_xlabel(r"Time $t$ [s]")
    ax.set_ylabel(r"Displacement $x$ [mm]")
````

## Inline PNG

The preview target sets the inline format and resolution, and optionally its size. A PNG preview keeps the notebook light when a figure has many points. A `Stdplot` binds the config to every figure drawn with it.

````python
png_config = Config(preview=Target(Format.PNG, Size(6.0, 4.0), dpi=150))
png_preview = Stdplot(png_config)

with png_preview.figure("damped_oscillation") as (fig, ax):
    ax.plot(time, displacement)
    ax.set_xlabel(r"Time $t$ [s]")
    ax.set_ylabel(r"Displacement $x$ [mm]")
````

Saving works as in a script: set `output_dir` in the config, then pass `save=True` once to the `Stdplot`, or to a single call, as in `export.py`.
